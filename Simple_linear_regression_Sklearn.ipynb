```python
# Import libraries
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
```

```python
#Dataset
X = np.array([1,2,3,4,5,6,7,8])
y = np.array([30,35,40,45,50,55,60,65])
```

```python
#Reshape X
X = X.reshape(-1,1)
```

```python
#Create and train the model
model = LinearRegression()
model.fit(X,y)
```

```python
#Get the model parameters
slope = model.coef_
intercept = model.intercept_
print(f"Slope: {slope}")
print(f"Intercept: {intercept:.2f}")
```

```python
#Make prediction
y_pred = model.predict(X)
print(y_pred)
```

```python
#predict a new value - predict salary for 10 years of experience
new_X = np.array([[10]])
prediction = model.predict(new_X)
print(f"Salary for 10 years: {prediction[0]:.2f}")
```

```python
#Model evaluation(R2)
r2 = model.score(X, y)
print(f"r2: {r2}")
```

```python
# Visualize the result
plt.scatter(X, y, label="Actual Salary", color="red")
plt.plot(X, y_pred, label="Regression Line", color="blue")

# Mark the prediction for 10 years
plt.scatter(
    new_X[0, 0],
    prediction[0],
    color="green",
    marker="*",
    s=200,
    label="Prediction for 10 Years"
)

plt.xlabel("Years of Experience")
plt.ylabel("Salary")
plt.title("Simple Linear Regression")
plt.legend()
plt.grid(True)
plt.show()
```
